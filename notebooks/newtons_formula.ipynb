Newton's Formula for approximating the roots of $f(x)$.

$$x_{n+1} = x_{n} - \frac{f(x_{n})}{f^{\prime}(x_{n})}$$

Substitute $f(x+n) = x_n^2 - S$ and $f^{\prime} (x_n) = 2 x_n$ to obtain:

$$
\begin{aligned}
x_{n+1} &= x_{n} - \frac{x_n^2 - S}{2x_{n}} \\
&= x_{n} - (-1) \times \frac{(S - x_n^2)}{2x_{n}} \\
&= x_{n} + \frac{(S - x_n^2)}{2x_{n}}
\end{aligned}
$$
